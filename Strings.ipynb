```python
def capitalEnds(s):
    words = s.split()
    s1 = ""

    for word in words:
        if len(word) == 1:
            word = word.upper()
        else:
            word = word[0].upper() + word[1:-1] + word[-1].upper()

        s1 = s1 + word + " "

    return s1.strip()

s = input("Enter your string = ")
print(capitalEnds(s))
```

```python
s = input("Enter your string = ")

def capital_ends(s):
    s1 = ""
    length = len(s)

    for i in range(length):
        if s[i] == " ":
            s1 += " "
        elif i == 0 or s[i-1] == " ":
            s1 += s[i].upper()
        elif i == length-1 or s[i+1] == " ":
            s1 += s[i].upper()
        else:
            s1 += s[i]

    return s1

print(capitalEnds(s))
```
