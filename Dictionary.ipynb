```python
score = {"Virat":100,"Sachin":110,"Rohit":264,"Rahul":199}

for i in score:
    print(i,score[i])

for i in score.keys():
    print(i)

for i in score.values():
    print(i)

for i,j in score.items():
    print(i,j)
```

```python
student = {101:{"Name":"Bhakti","Age":18,"Marks":{"Physics":90,"Maths":90,"Chemistry":90}},
          102:{"Name":"Viya","Age":18,"Marks":{"Physics":95,"Maths":97,"Chemistry":98}}}
print(student[102])
print(student[102]["Marks"])
print(student[102]["Marks"]["Maths"])
#apply items,keys,values in nested dictionary

for i in student:
    print(i,student[i]["Name"],student[i]["Marks"]["Chemistry"])

for i,j in student.items():
    print(i,j["Name"],j["Marks"]["Maths"])
```

```python
# methods of dictionary
#d.clear() - None
#d.get() - Value
#d.keys() - Value
#d.values() - Value
#d.items() - Dict
#d.update() - None
```

```python
# clear()

d = {"a":1}
b = d
b.clear()
print(d)

x = {"a":1}
y = x
x = {}
print(y)
```

```python
# copy() - shallow copy if changes are made in nested then changes are made in all
a = {"Name":"Virat","Score":[80,90]}
b = a.copy()
b["Name"] = "Rohit"
b["Score"][0] = 99
print(a)
print(b)
```

```python
# update
a = {"Name":"Virat","Age":18}
a.update({"Age":20,"City":"Delhi"})
print(a)

a.update(Department = "CSE",Year = 1)
print(a)

b = {"x":1}
b.update([("y",2),("z",3)])
print(b)
```

```python
x = {n:n**2 for n in range(1,6)}
print(x)

x = {n:n**2 for n in range(1,21) if n%2 == 0}
print(x)
```

```python
m = {"Virat":91,"Rohit":45,"Dhoni":70}
n = {name:("Pass" if marks >= 50 else "Fail") for name,marks in m.items()}
print(n)
```

```python
# get()

m = {"Virat":91,"Rohit":45,"Dhoni":70}
x = m.get("Virat")
print(x)
y = m.get("Pant") # default return value uf key is not there
print(y)
print(m.get("Gill",90))
print(m.get("Rohit",0))
```

```python
f = "banana"
x = {f.get:f.count(i) for i in f}
print(x)
```

```python
f = "banana"
ans = []
for i in f:
    ans = ans + [i,f.count(i)]
print(dict(ans))
```
