```python
#slicing
# l[start:stop(not included):step(cannot be zero)]
print(l[1:3])
print(l[:4])
print(l[::-1])
#print(l[::0]) ValueError: slice step cannot be zero
print(l[100:400]) #empty list
print(l[2:99]) #auto adjusts
print(l[-1:-4]) #empty list
```

```python
a = [10,20]
b = a #b is now an alias pointing to [10,20]
c = a.copy() # copy in a different path
d = a[:] # creates a shallow copy of 'a'
"""Because a[:] only copies the outer list, any inner lists inside a are still shared aliases. 
If a contains another list (like [[10, 20], 30]), modifying the inner list via d[0].append(99) will still affect a."""
a[0] = 99
print(b)
print(c)
print(d)
print(a is b)
print(a == b)
print(a is c)
print(a == c)
```

```python
#a.append() - to add value to the list

a = [1,2]
a.append(3)
print(a)

a.append([4,5])
print(a)

print(len(a))

x = a.append(6)
print(x) #append does not have the work to return so it returns none

print(a)
#a.append(7,8) TypeError: list.append() takes exactly one argument (2 given)
```

```python
#a.extend - modify list

a = [1,2]

a.extend([3,4])
print(a)

a.extend((5,6))
print(a)

a.extend("AB")
print(a)

#a.extend(10) TypeError: 'int' object is not iterable

a.extend([10])
print(a)
```

```python
# a.insert(index,value)

a = [10,20,30]
a.insert(1,5)
print(a)

a.insert(100,99)
print(a)
```

```python
# pop()

a = [10,20,30,40]
b = a.pop(0)
print(b)
print(a)

c = a.pop() # removes from last
print(a)
print(c)

# d = a.pop(-5) IndexError: pop index out of range
```

```python
# sort()
a = [1,5,9,3,8,4,7,4,8,0]
a.sort()
print(a)
a.sort(reverse = True)
print(a)

s = [1,1.0,True]
s.sort()
print(s)

s = [1,"Virat"]
#s.sort() TypeError: '<' not supported between instances of 'str' and 'int'

s = [True,"Virat"]
#s.sort() TypeError: '<' not supported between instances of 'str' and 'bool'

s = ["Thakur", "Sharma","Samay","Thakur"]
s.sort()
print(s)

w = ["pear","fig","banana"]
w.sort()
print(w)

w.sort(key = len)
print(w)

w.sort(key = len,reverse = True)
print(w)

w.sort(key = max)
print(w)
```

```python
#remove()

a = [10,20,10,30]
a.remove(10)
print(a)

print(a.remove(20))

#a.remove(10,20) TypeError: list.remove() takes exactly one argument (2 given)
#a.remove(100) ValueError: list.remove(x): x not in list
```

```python
# len, max, min, sum

a = [1,2,3,4,5,6,7]
print(len(a))
print(min(a))
print(max(a))
print(sum(a))
print(sum(a,100))
```

```python
# reverse()

a = [1,2,3] # if want to modify original use this
a.reverse()
print(a)
b = a.reverse()
print(b)

a = [10,20,30] # if want both original and reverse
b = a[::-1]
print(a)
print(b)
```

```python
# index

a = ["Jay","Viru","Gabbar","Basanti"]
print(a.index("Viru"))
#print(a.index("Vi")) ValueError: list.index(x): x not in list

b = ["Thakur","Dhanno","Sambha","Thakur"]
c = b.index("Thakur")
print(c)
d = b.index("Thakur",1) # 1 is starting index
print(d)
e = b.index("Thakur",0,3) # 0 is starting index and 3 is stopping index
print(e)
```

```python
# list comprehension - only a python feature

x = [n**2 for n in range(1,6)]
print(x)

y = [n for n in range(1,51) if n%2 == 0] #if n%2 == 0 is optional
print(y)

w = ["sachin","virat","rohit"]
w1 = [s.upper() for s in w]
print(w1)

a = [1,2,3,4]
ans = ["Even" if i%2==0 else "Odd" for i in a] #"Even" if i%2==0 else "Odd" - expression
print(ans)

#expression can be if-else,functions etc,followed by a for loop,
#then if required then and then only use if questions

p = [[1,2],[3,4]]
p1 = [i for j in p for i in j] 
print(p1)
```
